```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# Data for Bumrah's Format-Wise Performance
data_format = {
    'Format': ['Test', 'ODI', 'T20I'],
    'Matches Played': [45, 89, 70],
    'Wickets': [205, 149, 89],
    'Economy Rate': [2.76, 4.59, 6.27],
    'Bowling Average': [19.4, 23.55, 17.74],
}

# Create DataFrame for Format-Wise Performance
df_format = pd.DataFrame(data_format)

# Data for Bumrah's Performance against Oppositions
data_opposition = {
    'Opposition': ['Afghanistan', 'Australia', 'Bangladesh', 'England', 'Ireland', 'Namibia', 'Netherlands', 'New Zealand', 'Pakistan', 'Scotland',
                   'South Africa', 'Sri Lanka', 'U.A.E.', 'U.S.A.', 'West Indies', 'Zimbabwe'],
    'Matches Played': [4, 47, 11, 26, 4, 1, 1, 28, 12, 1, 23, 27, 1, 1, 11, 6],
    'Wickets': [10, 111, 27, 86, 8, 2, 2, 36, 12, 2, 58, 45, 1, 0, 29, 14],
    'Economy Rate': [3.92, 3.76, 3.87, 3.23, 4.26, 4.75, 3.66, 4.11, 4.72, 2.72, 3.37, 4.17, 5.75, 6.25, 3.65, 3.56],
    'Bowling Average': [11, 22.75, 16.03, 20.74, 8, 9.5, 16.5, 34.13, 25.66, 5, 21.7, 16.55, 23, 6.25, 12.48, 9.64],
}

# Create DataFrame for Opposition-Wise Performance
df_opposition = pd.DataFrame(data_opposition)

# Improved Visualizations with Seaborn and Plotly for Format-Wise and Opposition-Wise Analysis

# Format-Wise Visualization
# Plotting Wickets by Format using Seaborn
plt.figure(figsize=(8, 6))
sns.barplot(x='Wickets', y='Format', data=df_format, errorbar=None, palette=['#1f77b4', '#ff7f0e', '#2ca02c'])  # Custom Colors
plt.title('Wickets Taken by Bumrah in Different Formats', fontsize=16)
plt.xlabel('Wickets', fontsize=12)
plt.ylabel('Format', fontsize=12)
plt.tight_layout()
plt.show()

# Plotting Economy Rate by Format using Seaborn
plt.figure(figsize=(8, 6))
sns.barplot(x='Economy Rate', y='Format', data=df_format, errorbar=None, palette=['#9467bd', '#8c564b', '#e377c2'])  # Custom Colors
plt.title('Economy Rate by Bumrah in Different Formats', fontsize=16)
plt.xlabel('Economy Rate', fontsize=12)
plt.ylabel('Format', fontsize=12)
plt.tight_layout()
plt.show()

# Plotting Bowling Average by Format using Seaborn
plt.figure(figsize=(8, 6))
sns.barplot(x='Bowling Average', y='Format', data=df_format, errorbar=None, palette=['#7f7f7f', '#bcbd22', '#17becf'])  # Custom Colors
plt.title('Bowling Average by Bumrah in Different Formats', fontsize=16)
plt.xlabel('Bowling Average', fontsize=12)
plt.ylabel('Format', fontsize=12)
plt.tight_layout()
plt.show()

# Opposition-Wise Visualization
# Plotting Wickets by Opposition using Seaborn
plt.figure(figsize=(10, 8))
sns.barplot(x='Wickets', y='Opposition', data=df_opposition, errorbar=None, palette='tab20c')  # Using Seaborn 'tab20c' palette
plt.title('Wickets Taken by Bumrah Against Different Oppositions', fontsize=16)
plt.xlabel('Wickets', fontsize=12)
plt.ylabel('Opposition', fontsize=12)
plt.tight_layout()
plt.show()

# Plotting Economy Rate by Opposition using Seaborn
plt.figure(figsize=(10, 8))
sns.barplot(x='Economy Rate', y='Opposition', data=df_opposition, errorbar=None, palette='tab20b')  # Using Seaborn 'tab20b' palette
plt.title('Economy Rate by Bumrah Against Different Oppositions', fontsize=16)
plt.xlabel('Economy Rate', fontsize=12)
plt.ylabel('Opposition', fontsize=12)
plt.tight_layout()
plt.show()

# Plotting Bowling Average by Opposition using Seaborn
plt.figure(figsize=(10, 8))
sns.barplot(x='Bowling Average', y='Opposition', data=df_opposition, errorbar=None, palette='Set2')  # Using Seaborn 'Set2' palette
plt.title('Bowling Average by Bumrah Against Different Oppositions', fontsize=16)
plt.xlabel('Bowling Average', fontsize=12)
plt.ylabel('Opposition', fontsize=12)
plt.tight_layout()
plt.show()

# Combined Format and Opposition Insights
total_wickets_format = df_format['Wickets'].sum()
avg_economy_rate_format = df_format['Economy Rate'].mean()
avg_bowling_avg_format = df_format['Bowling Average'].mean()

total_wickets_opposition = df_opposition['Wickets'].sum()
avg_economy_rate_opposition = df_opposition['Economy Rate'].mean()
avg_bowling_avg_opposition = df_opposition['Bowling Average'].mean()

print("Combined Performance Insights:")
print(f"Total Wickets Taken Across All Formats: {total_wickets_format}")
print(f"Overall Economy Rate Across All Formats: {avg_economy_rate_format:.2f}")
print(f"Overall Bowling Average Across All Formats: {avg_bowling_avg_format:.2f}")
print(f"Total Wickets Taken Across All Oppositions: {total_wickets_opposition}")
print(f"Overall Economy Rate Across All Oppositions: {avg_economy_rate_opposition:.2f}")
print(f"Overall Bowling Average Across All Oppositions: {avg_bowling_avg_opposition:.2f}")
```
